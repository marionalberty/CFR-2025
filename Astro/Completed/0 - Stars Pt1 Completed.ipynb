## What is a star?
A star is a ball of gas and plasma (material that is very hot) that is held together by gravity. It gives off light and exists in space. Our Sun is the closest star to Earth.

## How do they form?
Stars form when really, really massive clouds of gas in space collapse. When these clouds collapse, gravity smushes the gas together. It forces the gas so closely together, that it creates a star.

## What are star made of?
Stars are made of different kinds of elements. An element is a fundamental building block of everything you use! For example, you might have heard of water being called "H2O," because it is made up of the elements Hydrogen and Oxygen.

Stars are mostly made of Hydrogen and Helium (what makes balloons float). Stars have other different kinds of elements as well. In fact, stars make most of the other elements we have on Earth. That cheesy saying "you're made of star stuff" is totally true! The Oxygen you're breathing now came from a star.

## How do stars give off light?
Stars give off an incredible amount of light. Some of this light is heat. That's why the Earth is warm! Our Sun sends heat and light we can see to Earth. All other stars also give off light. 

As mentioned before, stars are being crushed by gravity. Gravity is so strong that it forces elements together. For example, 1 Hydrogen is squished so strongly to another Hydrogen that it makes a different element, Helium! This process of making new elements be forcing others together is called fusion. And fusion gives off A LOT of energy. This energy comes from the center of a star and then goes out into space. 

Humans would love to be able to do fusions on Earth because it would be a great source of renewable energy. However, it takes a lot of energy to squish two elements together, and we humans cannot build machines that are as strong or effective as gravity. Maybe one day though....

Whenever we use data, it is important that we know both what the data is and what the units are for each variable. Below are the column names. We will **fill out** a definition for each column as we work through this notebook. First, let's identify the terms we might already know. 

| Column Header | Definition | Unit |
| :- | :- | :- |
| Star ID | identification number of the star | no unit

### Columns

We can look at a single column of data (one variable) at a time. To do this, we use the below structure:

data['Name of Column'], where you would fill in the name of the column exactly as it is in the data file. This can be a lot to type every time, so we can also save a whole column to a variable to save us some typing.

```python
# first, just print the temperatures
data['Temperature [K]']
```

```python
# now, save that column to a variable named "temperature"
temperature = data['Temperature [K]']
```

```python
# display the variable temperature to see that it is the same column
temperature
```

```python
# let's now create a variable for the columns Mass, Radius, and age
# make sure your variable names are one word and are useful (but not too long)
mass = data['Mass [Solar Mass]']
radius = data['Radius [Solar Radius]']
age = data['Age [Gyr]']
```

## Examining this population of stars

In this first data exploration, we will make histograms of the data we have and compare the stars Gaia observed to the sun. 

```python
# import our graphing package
import matplotlib.pyplot as plt

#the below is optional but makes the font a size I can read more easily
import matplotlib
matplotlib.rc('font', **{'family':'sans-serif', 'size':16}) 
```

```python
# create a histogram of the star temperatures. Make sure you label the axes
plt.hist(temperature)
plt.xlabel('Temperature [K]')
plt.ylabel('Number of Stars')
```

```python
# add a vertical line that represents the Sun's temperature at 5800 Kelvin
# you can do so with the line of code plt.avxline(), where you put the x number value in paraenthesis
# make sure you change the color of the line, so you can see it on top of the histogram

plt.hist(temperature)
plt.xlabel('Temperature [K]')
plt.ylabel('Number of Stars')
plt.axvline(5800, color='black')
```

**Roughly, how does the Sun's temperature compare to the population of stars? It is much hotter, much colder, similar, etc?**

**Answer: The Sun has a pretty average temperature, compared to the stars**

```python
# let's do the same thing now with the age
# the Sun's age is 4.6 Gyr

plt.hist(age)
plt.xlabel('Age [Gyr]')
plt.ylabel('Number of Stars')
plt.axvline(4.6, color='black')
```

**Roughly, how does the Sun's age compare to the population of stars? It is much younger, much older, similar, etc?**

**Answer: The Sun has a pretty average age, compared to the stars**

```python
# let's do the same thing now with the mass
# the Sun's mass is 1 Mo

plt.hist(mass)
plt.xlabel('Mass [Mo]')
plt.ylabel('Number of Stars')
plt.axvline(1, color='black')
```

**Roughly, how does the Sun's mass compare to the population of stars? It is much less massive, much more massive, similar, etc?**

**Answer: The Sun has a pretty average mass, compared to the stars**

```python
# let's do the same thing now with the radius
# the Sun's mass is 1 Mo

plt.hist(radius)
plt.xlabel('Radius [Ro]')
plt.ylabel('Number of Stars')
plt.axvline(1, color='black')
```

**Roughly, how does the Sun's radius compare to the population of stars? It is much smaller, much larger, similar, etc?**

**Answer: The Sun has a slightly smaller radius, compared to the stars**

**Looking at your 4 histograms, what conclusion can we make about the Sun? Is it pretty normal or pretty special?**

**Answer: The Sun is pretty normal!**
