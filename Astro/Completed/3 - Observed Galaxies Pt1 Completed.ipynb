## Deliverables: Do galaxies behave like stars?

Let's make some histograms to see if the mass, magnitudes, and colors are similar between stars and galaxies.

```python
# first, let's create variables for the columns GMAG, RMAG, and LOG_MASS
g = data['GMAG']
r = data['RMAG']
log_mass = data['LOG_MASS']
```

```python
# import our graphing package
import matplotlib.pyplot as plt

#the below is optional but makes the font a size I can read more easily
import matplotlib
matplotlib.rc('font', **{'family':'sans-serif', 'size':16}) 
```

```python
# histogram of g -- you might want to experiment with the number of bins (bins=#)
plt.hist(g, bins=50)
plt.ylabel('Number of Galaxies')
plt.xlabel('g [mag]')
```

**Look at your g mag histogram for stars. What are some differences and similarities you see?**

**Answer:** Both histograms have a clear peak. The magntidues for galaxies are lower than they are for stars.

**Does what you see make sense physically?**

**Answer:** Yes! Lower magnitudes means brighter. Since galaxies have billions of stars in them, they should be brighter than a single star.

```python
# histogram of r -- you might want to experiment with the number of bins (bins=#)
plt.hist(r, bins=50)
plt.ylabel('Number of Galaxies')
plt.xlabel('r [mag]')
```

**Look at your r mag histogram for stars. What are some differences and similarities you see?**

**Answer:** Both histograms have a clear peak. The magntidues for galaxies are lower than they are for stars. The galaxies also seem to have a cutoff at r > -20 so the shape is less like a bell and more asymmetrical.

**Does what you see make sense physically?**

**Answer:** Yes! Lower magnitudes means brighter. Since galaxies have billions of stars in them, they should be brighter than a single star.

```python
# create a variable for color and then create a histogram for it
# just like before, add labels for what is "redder" and what is "bluer"

color = g - r

plt.hist(color, bins = 100)
plt.xlim(0,1.25)
plt.axvline(0.6, color='black')
plt.annotate('bluer', (0.15, 5000), color='blue', fontsize=16)
plt.annotate('redder', (1, 5000), color='red', fontsize=16)
plt.xlabel('Color [g-r]')
plt.ylabel('Number of Galaxies')
```

**Look at your color histogram for stars. What are some differences and similarities you see?**

**Answer:** Both histograms are asymmetrical: they don't peak in the middle. Both have similar color value ranges even though the magnitudes had very different ranges. The stars are overall bluer than the galaxies.

**Does what you see make sense physically?**

**Answer:** Sure. It makes sense that the colors have similar values because colors don't add together the way brightness does.

### Mass

The mass is currently in log(Solar Mass) units. To compare directly to the stars, let's change it to just be Solar Mass.

Remember $log(A) = b$ and $10^b = A$ so $log(mass) = log \ mass$ and $10^{log\ mass} = mass$

Below, put $10^{log\ mass} = mass$   into code

```python
# solving for mass
mass = 10 ** log_mass
```

```python
# histogram of mass

plt.hist(mass, bins=100)
plt.xlim(0, 0.3e12)
plt.xlabel('Mass [Mo]')
plt.ylabel('Number of Galaxies')
```

**Look at your mass histogram for stars. What are some differences and similarities you see?**

**Answer:** Both histograms have a clear peak and more low mass objects than high mass. The ranges of mass are extremely different.

**Does what you see make sense physically?**

**Answer:** Yes! Since galaxies have billions of stars in them, they should be much more massive than a single star.
